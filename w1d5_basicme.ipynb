slicing #1

```python
first_name="muhammad"
last_name="ilyas"
full_name=first_name+" "+last_name
print(full_name)
lis=reversed(full_name)
print()
print(full_name[13:0:-1])
```

slicing #2

```python
fn=input("enter first name:")
ln=input("enter last name:")
print("first name=",fn)
print("last name=",ln)
rfn=fn[-1:-6:-1]
print(rfn)
rln=ln[::-1]
print(rln)
print(-len(fn))
```

Easy method

```python
fn=input("enter first name:")
ln=input("enter last name:")
print(fn)
print(ln)
rfn=fn[-1:-len(fn)-1:-1]
rln=ln[-1:-len(ln)-1:-1]
print(rfn)
print(rln)
```
