## while loop

when condition are given at the last while the value is stored in an initialized variable

```python
cor_uname="admin"
cor_pass="admin134"

maxattempt=3
attempt =0
while attempt <maxattempt:
    userna=input("enter user name:")
    pas=input("enter password:")
    attempt+=1
    if userna==cor_uname and cor_pass==pas:
        print(f"login successfuly!")
        break
    else:
        print(f"{userna} or {pas} is incorrect!")
        print(f"you have {maxattempt-attempt} attempt left")
```
