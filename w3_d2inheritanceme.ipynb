# inheritance

```python
class PYTHON:
    interest_rate=0.5
    branch='cs technology'
    def __init__(self,owner,amount):
        self.owner =owner
        self.balance=amount

    def deposit(self,amount):
        self.balance+=amount
    def change_interestrate(cls,value):
        cls.interest_rate=value
    @staticmethod
    def validate_amount(amount):
        if amount<=0:
            return False
        elif amount>100000:
            return False
        else:
            return True
```

```python
p1=PYTHON('ahmad',-1000)
```

## encapsulation

```python
class Student:
    def __init__(self,name,age,gpa):
        self.name=name
        self.age=age
        self.gpa=gpa
        self.courses=[]
    def show_detail(self):
        print(f"-------Student detail----------")
        print(f"{self.name}")
        print(f"{self.age}")
        print(f"{self.gpa}")
    
```

```python
s1=Student("Ahmad",31,4)
```

```python
s1.name
```

```python
class Student:
    def __init__(self,name,age,gpa):
        self.name=name
        self.age=age
        self.__gpa=gpa
        self.courses=[]
    def show_detail(self):
        print(f"-------Student detail----------")
        print(f"{self.name}")
        print(f"{self.age}")
        print(f"{self.__gpa}")
```

```python
s1=Student("Ahmad",31,4)
```

```python
s1.name
```

```python
s1.gpa
```
