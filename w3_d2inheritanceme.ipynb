```python
class PYTHON:
    interest_rate=0.5
    branch='cs technology'
    def __init__(self,owner,amount):
        self.owner =owner
        self.balance=amount

    def deposit(self,amount):
        self.balance+=amount
    def change_interestrate(cls,value):
        cls.interest_rate=value
    @staticmethod
    def validate_amount(amount):
        if amount<=0:
            return False
        elif amount>100000:
            return False
        else:
            return True
```

```python
class Student:
    def __init__(self,name,age,gpa):
        self.name=name
        self.age=age
        self.__gpa=gpa
        self.courses=[]
    def show_detail(self):
        print(f"-------Student detail----------")
        print(f"{self.name}")
        print(f"{self.age}")
        print(f"{self.__gpa}")
```

```python
class point:
    def __init__(self,x,y):
        self.__x=x
        self.__y=y

    @property
    def x(self):
        return self.__x
    
    @x.setter
    def x(self,value):
        self.__x=value

    @property
    def y(self):
        return self.__y
    
    @y.setter
    def y(self,value):
        self.__y=value

    def __str__(self):
        return f"{self.__x}, {self.__y}"
    
    def __add__(self, other):
        return self.x + other.x
    
    def calculatexy(self):
        return (((self.x**2)+(self.y**2))**0.5)
```

```python
point1=point(9,1)
```

```python
point1.x
```

```python
point1.y
```

```python
point1.calculatexy()
```
