```python
X_eng = np.c_[X[:, 0], X[:, 0]**2, X[:, 1], X[:, 2], X[:, 2]**2, X[:, 3], X[:, 4], X[:, 5]]
```

```python
p_model = LinearRegression()

p_model.fit(X_eng, y)
```

```python
pred = p_model.predict(X_eng)
```
