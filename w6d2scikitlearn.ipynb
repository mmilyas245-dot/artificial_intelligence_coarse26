```python
from sklearn.linear_model import LinearRegression

import pandas as pd
import numpy as np
import matplotlib as plt

from sklearn.model_selection import train_test_split

from sklearn.metrics import r2_score,root_mean_squared_error
```

```python
df=pd.read_csv(r"C:\artificial_intelligence_coarse\kc_house_data.csv")
```

```python
df.head(5)
```

```python
df.size
```

```python
df.info()
```

```python
df=df.drop(columns=['id','date'])
```

```python
df.info()
```

```python
df.corr()
```

```python
Y=df['price']
```

```python
df.corr()['price'].sort_values(ascending=False).drop('price')
```

```python
feature=['sqft_living','grade','sqft_above','sqft_living15','bathrooms','view','bedrooms','lat','waterfront','floors']
```

```python
X = df[feature]
```

```python
x_train,x_test,y_trian,y_test=train_test_split(X,Y,test_size=0.2,random_state=(42))
```

```python
house_price=LinearRegression()
```

```python
house_price.fit(x_train,y_trian)
```

```python
y_prediction=house_price.predict(x_test)
```

```python
len(df)*0.2
```

```python
y_test
```

```python
r2_score(y_true=y_test,y_pred=y_prediction)
```

```python
root_mean_squared_error(y_test,y_prediction)
```

```python
house_price.intercept_
```
