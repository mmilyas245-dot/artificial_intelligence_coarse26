one name but found in many shape is known as method overriding

one name but can take two parameter in one function while taking three parameter in another place as known as method overloading and operator overloading

```python
class shape:
    def area(self):
        return self.pi * self.radius **2
class Circle(shape):
    def __init__(self,r):
        self.radius=r
        self.pi=3.14
class Rectangle(shape):
    def __init__(self,l,w):
        self.width=w
        self.length=l
```

# Abstraction

Hiding internal detail

hiding unnecesssary detail

```python
from abc import ABC,abstractmethod
class shape(ABC):
    def area(self):
        return 0
    def perimeter(self):
        return 0
    @abstractmethod
    def describe (self):
        print(f"Area : {self.area()}  |  Perimeter : {self.perimeter()}")
```

```python
c1=shape()
```

```python
c1.describe()
```

```python
c1.describe()
```

```python
c1.area()
```

```python
class Counter:
    def __init__(self,value=0):
        self.value=value
    def increment(self):
        self.value += 1
        return f"Value incremented by 1: {self.value}"
    
    def dectrement(self):
        if self.value>0:
            if self.value==1:
                self.value -=1
                return f"Warning decrement value! :{self.value-1}"
        
            return f"Value decremented by 1: {self.value-1}"
        else:
            print("value must be greater than zero!")

    def reset(self):
        return self.value ==0
```

```python
c1=Counter(5)
```

```python
c1.value
```

```python
c1.dectrement()
```

```python
c1.increment()
```

```python
c1.reset=1
```

```python
c1.reset
```

```python
class vehicle:
    def __init__(self,brand,wheels):
        self.brand=brand
        self.wheels=wheels
class motorcycle(vehicle):
    def __init__(self, brand, wheels,has_helmet_storage):
        super().__init__(brand, wheels)
        self.has_helmet_storage=has_helmet_storage
```
