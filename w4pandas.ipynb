```python
import matplotlib.pyplot as plt

# 🔢 Your data (you can change these values)
values = [len(pstd),len(lahorestd)]

# 🏷️ Labels for each slice
labels = ['Peshawar', 'Lahore']#, 'Karachi', 'Islamabad', 'Multan']

# 🎨 Colors for each slice (you can customize)
colors = ['gold', 'violet']#, 'lightskyblue', 'lightgreen', 'violet']

# 🚀 Create the pie chart
plt.pie(
    values,                 # data values
    labels=labels,          # slice labels
    colors=colors,          # slice colors
    autopct='%1.1f%%',      # show percentage (1 decimal)
    startangle=140          # rotate chart for better view
)

# 🔵 Equal aspect ratio ensures circle shape
plt.axis('equal')

# 🏷️ Title of chart
plt.title("Programming Language Usage")

# 📊 Show the chart
plt.show()
```
